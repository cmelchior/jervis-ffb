# Generate Serializer Module

The Jervis Serializer Module is pretty big and contains a lot of classes. This notebook will generate the module from parsing the classpath and find all relevant sub-classes.

```kotlin
// implementation("org.jetbrains.kotlin:kotlin-reflect")
@file:DependsOn("io.github.classgraph:classgraph:4.8.157")
```

```kotlin
import kotlin.reflect.KClass

// Helper class to track the type hiearchy
class SerializedClass(
    val type: KClass<*>,
    val finalSubTypes: MutableList<KClass<*>> = mutableListOf(),
    val openSubTypes: MutableList<SerializedClass> = mutableListOf()
)
```

```kotlin
import com.jervisffb.engine.actions.CalculatedAction
import com.jervisffb.engine.actions.GameAction
import com.jervisffb.engine.model.inducements.BiasedReferee
import com.jervisffb.engine.model.inducements.Spell
import com.jervisffb.engine.model.inducements.wizards.Wizard
import com.jervisffb.engine.model.inducements.InfamousCoachingStaff
import com.jervisffb.engine.model.locations.Location
import com.jervisffb.engine.rules.Rules
import  com.jervisffb.engine.rules.builder.BallSelectorRule
import  com.jervisffb.engine.rules.builder.StadiumRule
import com.jervisffb.engine.rules.common.procedures.DieRoll
import com.jervisffb.engine.rules.common.actions.TeamActions
import com.jervisffb.engine.rules.common.skills.RerollSource
import com.jervisffb.engine.rules.common.skills.Skill
import com.jervisffb.engine.rules.common.skills.SkillSettings
import com.jervisffb.engine.rules.common.tables.ArgueTheCallTable
import com.jervisffb.engine.rules.common.tables.CasualtyTable
import com.jervisffb.engine.rules.common.tables.InjuryTable
import com.jervisffb.engine.rules.common.tables.KickOffTable
import com.jervisffb.engine.rules.common.tables.LastingInjuryTable
import com.jervisffb.engine.rules.common.tables.PrayersToNuffleTable
import com.jervisffb.engine.rules.common.tables.RangeRuler
import com.jervisffb.engine.rules.common.tables.WeatherTable
import com.jervisffb.engine.rules.common.pathfinder.PathFinder
import com.jervisffb.engine.rules.common.roster.Position
import com.jervisffb.engine.rules.common.roster.Roster
import com.jervisffb.engine.serialize.SerializedPlayer
import com.jervisffb.engine.serialize.SerializedTeam
import com.jervisffb.engine.model.inducements.settings.Inducement
import io.github.classgraph.ClassGraph
import kotlin.reflect.KClass

// List of top-level interfaces and abstract classes that can be
// serialized
// Do not serialize Procedures and ProcedureContex's. They should
// be regenerated when loading initial actions.
val topLevelClasses = mutableListOf<KClass<*>>(
    ArgueTheCallTable::class,
    BallSelectorRule::class,
    BiasedReferee::class,
    CasualtyTable::class,
    DieRoll::class,
    GameAction::class,
    Inducement::class,
    InfamousCoachingStaff::class,
    InjuryTable::class,
    KickOffTable::class,
    LastingInjuryTable::class,
    PathFinder::class,
    Position::class,
    PrayersToNuffleTable::class,
    RangeRuler::class,
    Rules::class,
    SerializedPlayer::class,
    SerializedTeam::class,
    SkillSettings::class,
    StadiumRule::class,
    Spell::class,
    TeamActions::class,
    WeatherTable::class,
    Wizard::class,
)

// Some subtypes we explicetely do not want to serialize.
// List them here.
val ignoredClasses: Set<KClass<*>> = setOf(
    CalculatedAction::class
)

// Create ClassGraph object we can use to iterate over the class hieachy.
val classGraph = ClassGraph()
    .enableAllInfo()
    .acceptPackages("com.jervisffb.engine")
    .enableInterClassDependencies()
    .scan()

fun visitOpenClass(clazz: KClass<*>, serializeInfo: SerializedClass) {
    val subClasses = if (clazz.java.isInterface) {
        classGraph
            .getClassesImplementing(clazz.java)
            .map { Class.forName(it.name).kotlin }
            .filterNot { it: KClass<*> -> ignoredClasses.contains(it) }
    } else {
        classGraph
            .getSubclasses(clazz.java)
            .map { Class.forName(it.name).kotlin }
            .filterNot { it: KClass<*> -> ignoredClasses.contains(it) }
    }
    for (subClass: KClass<*> in subClasses) {
        if (subClass.java.isInterface || subClass.isAbstract || subClass.isSealed) {
            val newTypeInfo = SerializedClass(subClass)
            serializeInfo.openSubTypes.add(newTypeInfo)
            visitOpenClass(subClass, newTypeInfo)
        } else {
            serializeInfo.finalSubTypes.add(subClass)
        }
    }
}

// Build up the class hiearchy
val serializableHierachy: List<SerializedClass> = topLevelClasses.map { serializableType: KClass<*> ->
    val info = SerializedClass(serializableType)
    visitOpenClass(serializableType, info)
    info
}
```

```kotlin
// Create three serializer module files: core (common-only), bb2020, bb2025.
// Each polymorphic base's subclass list is bucketed by the fully-qualified
// name of the subclass. Common subclasses go into the core file; BB2020/BB2025
// subclasses go into the corresponding ruleset file. If a bucket ends up
// empty for a given base, the polymorphic block is omitted from that file.

fun bucketFor(fqn: String?): String = when {
    fqn == null -> "core"
    fqn.startsWith("com.jervisffb.engine.rules.bb2020.") -> "bb2020"
    fqn.startsWith("com.jervisffb.engine.rules.bb2025.") -> fqn.let {
        if (it == "com.jervisffb.engine.rules.bb2025rules.kt") "bb2025" else "bb2025"
    }
    else -> "core"
}

// The concrete BB2020Rules / BB2025Rules subclasses live in
// `com.jervisffb.engine.rules.BB72020Rules` etc. (top-level `rules` package).
// Route them to their ruleset bucket by class simple-name prefix as a fallback.
fun bucketForRulesClass(fqn: String?): String? = when {
    fqn == null -> null
    fqn.startsWith("com.jervisffb.engine.rules.BB2020") -> "bb2020"
    fqn.startsWith("com.jervisffb.engine.rules.StandardBB2020") -> "bb2020"
    fqn.startsWith("com.jervisffb.engine.rules.FumbblBB2020") -> "bb2020"
    fqn.startsWith("com.jervisffb.engine.rules.BB72020") -> "bb2020"
    fqn.startsWith("com.jervisffb.engine.rules.BB2025") -> "bb2025"
    fqn.startsWith("com.jervisffb.engine.rules.StandardBB2025") -> "bb2025"
    else -> null
}

fun classifySubclass(fqn: String?): String {
    return bucketForRulesClass(fqn) ?: bucketFor(fqn)
}

// Abstract Rules subclasses (BB2020Rules, BB2025Rules) are NOT @Serializable
// (they'd trigger a Kotlin 2.4 cross-module SyntheticAccessorLowering bug), so
// we skip generating nested `polymorphic(BB2020Rules::class) { ... }` blocks.
// Concrete subclasses (StandardBB2020Rules etc.) use custom KSerializers
// defined next to each class; we emit `subclass(Foo::class, FooSerializer)`.
val nonSerializableAbstractRules: Set<String> = setOf(
    "com.jervisffb.engine.rules.BB2020Rules",
    "com.jervisffb.engine.rules.BB2025Rules",
)

fun customRulesSerializerFor(fqn: String?): String? = when (fqn) {
    "com.jervisffb.engine.rules.StandardBB2020Rules" -> "com.jervisffb.engine.rules.StandardBB2020RulesSerializer"
    "com.jervisffb.engine.rules.FumbblBB2020Rules" -> "com.jervisffb.engine.rules.FumbblBB2020RulesSerializer"
    "com.jervisffb.engine.rules.BB72020Rules" -> "com.jervisffb.engine.rules.BB72020RulesSerializer"
    "com.jervisffb.engine.rules.StandardBB2025Rules" -> "com.jervisffb.engine.rules.StandardBB2025RulesSerializer"
    else -> null
}

data class BucketedNode(
    val type: SerializedClass,
    // finalSubTypes bucketed
    val core: MutableList<KClass<*>> = mutableListOf(),
    val bb2020: MutableList<KClass<*>> = mutableListOf(),
    val bb2025: MutableList<KClass<*>> = mutableListOf(),
    val nested: MutableList<BucketedNode> = mutableListOf()
)

fun bucketize(type: SerializedClass): BucketedNode {
    val node = BucketedNode(type)
    for (sub in type.finalSubTypes) {
        when (classifySubclass(sub.qualifiedName)) {
            "bb2020" -> node.bb2020.add(sub)
            "bb2025" -> node.bb2025.add(sub)
            else -> node.core.add(sub)
        }
    }
    for (openSub in type.openSubTypes) {
        // Skip nested polymorphic blocks for abstract Rules subclasses
        // that are no longer @Serializable — the concrete children are
        // still registered under the direct `Rules::class` polymorphic.
        if (openSub.type.qualifiedName in nonSerializableAbstractRules) continue
        node.nested.add(bucketize(openSub))
    }
    return node
}

fun BucketedNode.hasContent(bucket: String): Boolean = when (bucket) {
    "core" -> core.isNotEmpty() || nested.any { it.hasContent("core") }
    "bb2020" -> bb2020.isNotEmpty() || nested.any { it.hasContent("bb2020") }
    "bb2025" -> bb2025.isNotEmpty() || nested.any { it.hasContent("bb2025") }
    else -> false
}

fun subclassesFor(node: BucketedNode, bucket: String): List<KClass<*>> = when (bucket) {
    "core" -> node.core
    "bb2020" -> node.bb2020
    "bb2025" -> node.bb2025
    else -> emptyList()
}

fun StringBuilder.emitPolymorphic(node: BucketedNode, bucket: String, indent: Int) {
    if (!node.hasContent(bucket)) return
    val topIndent = " ".repeat(indent * 4)
    val nestedIndent = " ".repeat((indent + 1) * 4)
    appendLine("${topIndent}polymorphic(${node.type.type.qualifiedName}::class) {")
    for (sub in subclassesFor(node, bucket)) {
        val serializer = customRulesSerializerFor(sub.qualifiedName)
        if (serializer != null) {
            appendLine("${nestedIndent}subclass(${sub.qualifiedName}::class, ${serializer})")
        } else {
            appendLine("${nestedIndent}subclass(${sub.qualifiedName}::class)")
        }
    }
    for (nested in node.nested) {
        emitPolymorphic(nested, bucket, indent + 1)
    }
    appendLine("$topIndent}")
}

val bucketedHierarchy: List<BucketedNode> = serializableHierachy.map { bucketize(it) }

fun buildFile(bucket: String, packageName: String, valName: String, kdoc: String): String = buildString {
    appendLine("""
        package $packageName

        import kotlinx.serialization.modules.SerializersModule
        import kotlinx.serialization.modules.polymorphic
        import kotlinx.serialization.modules.subclass

        /**
         * WARNING: This file is autogenerated. Do not modify manully
         * Instead rerun the notebook in `tools/GenerateSerializers.ipynb` to update.
         *
         * $kdoc
         */
        val $valName = SerializersModule {
    """.trimIndent())
    for (node in bucketedHierarchy) {
        emitPolymorphic(node, bucket, 1)
    }
    appendLine("}")
}

val coreCode = buildFile(
    "core",
    "com.jervisffb.engine.serialize",
    "coreSerializerModule",
    "Polymorphic subclass registrations for types whose entire class hierarchy lives in `engine.*` or `rules.common.*`."
)
val bb2020Code = buildFile(
    "bb2020",
    "com.jervisffb.engine.rules.bb2020.serialize",
    "bb2020SerializerModule",
    "BB2020 subclass registrations for polymorphic bases defined in `rules.common.*`."
)
val bb2025Code = buildFile(
    "bb2025",
    "com.jervisffb.engine.rules.bb2025.serialize",
    "bb2025SerializerModule",
    "BB2025 subclass registrations for polymorphic bases defined in `rules.common.*`."
)

listOf("core" to coreCode, "bb2020" to bb2020Code, "bb2025" to bb2025Code)
```
